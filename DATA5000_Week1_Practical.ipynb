## 3. Read it like a sceptic

Here is the real work of the session. Whatever the agent (or the summary above) told you, it almost certainly made **at least two confident moves it could not actually justify from the data alone.**

Look closely at two things.


```python
# Clue 1 — the contract_status column. What do these codes mean?
mobtel['contract_status'].value_counts()
```

The agent will have treated `contract_status` as meaningful, but the codes are **A**, **S** and **C**. Nothing in the data says what they stand for. An AI will often guess — perhaps "S means subscribed", perhaps it ignores the column. Only someone who knows the business knows the truth:

- **A** = Active
- **S** = Suspended (a payment problem — still a customer, not yet gone)
- **C** = Closed (this customer has actually left)

That distinction matters enormously. If you are predicting who will leave, treating a Suspended customer the same as a Closed one would be a real error — and the agent had no way to know the difference.


```python
# Clue 2 — the customers paying $0 a month. Who are they?
mobtel[mobtel['monthly_charge_aud'] == 0].head(12)
```

There are twelve accounts paying nothing. An AI cleaning this data will often quietly **drop them** as errors, or leave them in as if they were ordinary customers. Both could be wrong. In reality these are **staff and comped accounts** — real people using the network, but not paying customers. Whether they belong in a churn analysis is a *business* decision, and the agent made it silently, one way or the other, without telling you.
